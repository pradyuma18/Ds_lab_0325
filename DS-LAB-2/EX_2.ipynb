## Step 6  Option 1: Identifying Outliers Using Z-Scores
The value of 3 in the context of Z-scores is often used as a threshold to identify outliers in a dataset. A Z-score represents how many standard deviations a data point is away from the mean of the dataset. Specifically:

A Z-score of 0 means the data point is exactly at the mean.
A Z-score of 1 means the data point is one standard deviation above the mean, and so on.
A Z-score of 3 corresponds to a data point being 3 standard deviations away from the mean. For a normal distribution, about 99.7% of the data points fall within 3 standard deviations of the mean (according to the 68-95-99.7 rule, which describes the spread of data in a normal distribution). Therefore, points with Z-scores greater than 3 or less than -3 are considered unusually far from the mean and are often flagged as outliers.

This threshold (Z > 3 or Z < -3) is commonly used in many statistical applications because it captures the extreme values that are rare in a normal distribution, which are typically considered to be outliers. However, the choice of threshold can vary depending on the specific application and the nature of the data.

### 4. Handle Missing Values
`horsepower` is handled separately in Step 6, so it is skipped here.

```python
# Replace missing numerical values with the mean (skip horsepower, handled in Step 6)
for col in auto_df.select_dtypes(include=['int64', 'float64']).columns:
    if col == 'horsepower':
        continue
    if auto_df[col].isnull().sum() > 0:
        auto_df[col] = auto_df[col].fillna(auto_df[col].mean())
        print(f"Filled missing values in '{col}' with mean")

# Replace missing categorical values with the mode
for col in auto_df.select_dtypes(include=['object', 'str']).columns:
    if auto_df[col].isnull().sum() > 0:
        mode_val = auto_df[col].mode()[0]
        auto_df[col] = auto_df[col].fillna(mode_val)
        print(f"Filled missing values in '{col}' with mode: {mode_val}")
```

**Label Encoding vs. One-Hot Encoding for `origin`**

| Aspect | Label Encoding | One-Hot Encoding |
|---|---|---|
| Output | Single column of integers (0, 1, 2) | One binary column per category |
| Implies order? | Yes — implicitly suggests `europe < japan < usa` | No — treats categories as independent |
| Best for | Ordinal categories, or tree-based models | Nominal (unordered) categories, linear/distance-based models |

**Which is more appropriate for `origin`?**
`origin` is a **nominal** categorical variable — `usa`, `europe`, and `japan` have no inherent
rank or order. Label Encoding would incorrectly imply an ordinal relationship, which can
mislead distance-based or linear models. **One-Hot Encoding is the more appropriate choice**
here, since it avoids introducing a false ordinal relationship. Label Encoding is generally
acceptable only for tree-based models (Decision Trees, Random Forests, Gradient Boosting),
which don't assume any numeric ordering.

## Step 5: Outlier Detection
### 15. Identify Outliers Using Z-Scores

```python
numeric_features = ['miles_per_gallon', 'cylinders', 'displacement', 'hp',
                     'weight_lbs', 'acceleration', 'weight_kg']

z_scores = auto_df[numeric_features].apply(stats.zscore)
z_scores.columns = [f"{c}_zscore" for c in numeric_features]

z_scores.head()
```

```python
outlier_counts = (z_scores.abs() > 3).sum()
print("Outlier counts per numerical column:")
print(outlier_counts)
```

```python
outlier_mask = (z_scores.abs() > 3).any(axis=1)
outlier_rows = auto_df[outlier_mask]

print(f"Total rows containing at least one outlier: {len(outlier_rows)}")
outlier_rows
```

**Should the outliers be removed or retained?**

These extreme values (e.g. very high horsepower/displacement muscle cars, or very
light/high-mpg economy cars) are legitimate real-world data points rather than data-entry
errors. The recommendation is to **retain** them, but consider robust models (tree-based
methods), a log-transform for heavily skewed features, or a RobustScaler if their influence
during scaling is a concern. Removing them would discard genuine variation in the data.

**Standardization vs. Normalization**

| Aspect | Standardization (StandardScaler) | Normalization (MinMaxScaler) |
|---|---|---|
| Formula | `(x - mean) / std` | `(x - min) / (max - min)` |
| Resulting range | Unbounded (mean 0, std 1) | Fixed range `[0, 1]` |
| Sensitive to outliers | Less sensitive | Very sensitive (min/max driven) |
| Good for | Linear/logistic regression, SVM, PCA, k-means | Neural networks, KNN, bounded-input algorithms |

**When to use each:** Standardization suits roughly Gaussian data and algorithms assuming
zero-centered input. Normalization suits algorithms that need bounded input (e.g. neural
networks, KNN) but is more easily skewed by extreme outliers since it relies on min/max.
